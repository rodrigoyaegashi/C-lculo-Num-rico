**Questão 5 — Implemente o método de Gauss–Seidel em Python**

```python


x, y, z = 0.0, 0.0, 0.0
tolerancia = 1e-5
k = 0

while True:
    x_ant, y_ant, z_ant = x, y, z

    x = (6 + y_ant - 2*z_ant) / 10
    y = (22 + x + z_ant) / 11
    z = (-10 - 2*x + y) / 10

    erro = max(
        abs(x - x_ant),
        abs(y - y_ant),
        abs(z - z_ant)
    )

    k += 1

    if erro < tolerancia:
        break

print("Numero de iteracoes:", k)
print("Solucao aproximada:")
print(f"x = {x:.10f}")
print(f"y = {y:.10f}")
print(f"z = {z:.10f}")
print(f"Erro final = {erro:.10e}")
```

**Questão 6 — Tabela das iterações**

```python

import pandas as pd

x, y, z = 0.0, 0.0, 0.0
tolerancia = 1e-5
resultados = []
k = 0

while True:
    x_ant, y_ant, z_ant = x, y, z

    x = (6 + y_ant - 2*z_ant) / 10
    y = (22 + x + z_ant) / 11
    z = (-10 - 2*x + y) / 10

    erro = max(
        abs(x - x_ant),
        abs(y - y_ant),
        abs(z - z_ant)
    )

    k += 1
    resultados.append([k, x, y, z, erro])

    if erro < tolerancia:
        break

tabela = pd.DataFrame(
    resultados,
    columns=["k", "x(k)", "y(k)", "z(k)", "Erro"]
)

display(tabela.round(10))
```

**Questão 7 — Gráficos**

```python

import matplotlib.pyplot as plt

x, y, z = 0.0, 0.0, 0.0
tolerancia = 1e-5

ks = []
xs = []
ys = []
zs = []
erros = []

k = 0

while True:
    x_ant, y_ant, z_ant = x, y, z

    x = (6 + y_ant - 2*z_ant) / 10
    y = (22 + x + z_ant) / 11
    z = (-10 - 2*x + y) / 10

    erro = max(
        abs(x - x_ant),
        abs(y - y_ant),
        abs(z - z_ant)
    )

    k += 1

    ks.append(k)
    xs.append(x)
    ys.append(y)
    zs.append(z)
    erros.append(erro)

    if erro < tolerancia:
        break

# Grafico A: valores das incognitas
plt.figure(figsize=(8, 5))
plt.plot(ks, xs, marker="o", label="x")
plt.plot(ks, ys, marker="s", label="y")
plt.plot(ks, zs, marker="^", label="z")

plt.xlabel("Numero da iteracao")
plt.ylabel("Valor")
plt.title("Metodo de Gauss-Seidel")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# Grafico B: erro em escala logaritmica
plt.figure(figsize=(8, 5))
plt.semilogy(ks, erros, marker="o")

plt.xlabel("Numero da iteracao")
plt.ylabel("Erro maximo")
plt.title("Erro entre iteracoes")
plt.grid(True, which="both")
plt.tight_layout()
plt.show()
```
